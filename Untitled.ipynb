```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
y = x **3 
y
```

```python
plt.plot(x, y)
plt.title("Our plot")
plt.xlabel("Sell")
plt.ylabel("iteam")
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y)
```

```python
fig = plt.figure(figsize=(12,6),dpi =100)
axis1 = fig.add_axes([0.1,0.1,.9,.9]) # [_,_,width, height]
axis1.plot(x,y)
axis1 = fig.add_axes([0.4,0.4,.4,.4]) # [_,_,width, height]
axis1.plot(x,y)
```
