```python
#gender is a qualitative variable and age is a quantitative variable.
import seaborn as sns
import matplotlib.pyplot as plt 
```

We use distribution plots to visualize the distribution of quantitative data.

Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:

* distplot/histplot
* jointplot
* pairplot
* rugplot

```python
df = sns.load_dataset('tips')
df
```

```python
df['size'].unique()
```

```python
sns.histplot(df['total_bill'],bins = 20,kde = True) #kernel density estimation (KDE)
```

```python
# how to create muntiple plot in hist plot....
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins = 20,kde = True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins = 20,kde = True)
```

**Joint Plot**

we use joint plot to create or compare muntiple columns from the dataset

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df,kind = 'scatter')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df,kind = 'reg')
```

**pairplot**

```python
sns.pairplot(df,hue='sex') # hue is basically gving color to catagory
```

```python
sns.pairplot(df,hue='size',palette='rainbow')
```

```python
sns.rugplot(df['tip'])
```
